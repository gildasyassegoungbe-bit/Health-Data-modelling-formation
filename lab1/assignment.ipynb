Q4 : Define and comment on the evaluation metrics you need to select the best model. Justify your answer rigorously with examples.

Answer : ....

For this multiclass classification problem, several evaluation metrics are useful to assess model performance. Accuracy measures the proportion of correctly classified observations among all observations. Precision measures, for a given class, the proportion of correct predictions among all observations predicted as belonging to that class. Recall measures the proportion of correctly identified observations among all observations that actually belong to that class. The F1-score combines precision and recall and provides a balance between them. Finally, the confusion matrix shows the distribution of correct predictions and misclassifications between the different classes.

Using several metrics is important because accuracy alone does not describe the types of errors made by a multiclass model. For example, if 100 observations are truly classified as obese, and the model correctly identifies 70 of them, the recall for the obese class is 70%. If the model predicts 140 observations as obese and only 70 are truly obese, the precision is 50%. Therefore, precision and recall provide complementary information about model performance.

Q16 : Which attribute should be chosen for the root node?

```python
obesity_binary["ObesityCategory"].value_counts()
```

```python
import math

def entropy(y):
    counts = y.value_counts()
    total = len(y)
    
    ent = 0
    for count in counts:
        p = count / total
        ent -= p * math.log2(p)
        
    return ent


def information_gain(data, feature, target):
    total_entropy = entropy(data[target])
    
    weighted_entropy = 0
    
    for value in data[feature].unique():
        subset = data[data[feature] == value]
        weight = len(subset) / len(data)
        weighted_entropy += weight * entropy(subset[target])
    
    return total_entropy - weighted_entropy
```

```python
target = "ObesityCategory"

features = [
    "Age",
    "Gender",
    "Height",
    "Weight",
    "BMI",
    "PhysicalActivityLevel"
]

information_gains = {}

for feature in features:
    information_gains[feature] = information_gain(
        obesity_binary,
        feature,
        target
    )

information_gains
```

```python
for feature, gain in information_gains.items():
    print(f"{feature}: {gain:.4f}")
```

The attribute chosen for the root node is Weight, because it has the highest Information Gain (0.9544). BMI has the same Information Gain (0.9544), so the two attributes are tied. With the current feature order, Weight is selected first.

Q17 : Write a function to build the tree and visualize the tree.

```python
def build_id3_tree(data, features, target):
    
    # Stop 1: all observations belong to the same class
    if data[target].nunique() == 1:
        return data[target].iloc[0]
    
    # Stop 2: no feature remains
    if len(features) == 0:
        return data[target].mode()[0]
    
    # Calculate Information Gain for each feature
    information_gains = {}
    
    for feature in features:
        information_gains[feature] = information_gain(
            data,
            feature,
            target
        )
    
    # Choose the feature with the highest Information Gain
    best_feature = max(
        information_gains,
        key=information_gains.get
    )
    
    # Create the tree node
    tree = {best_feature: {}}
    
    # Remaining features
    remaining_features = [
        feature for feature in features
        if feature != best_feature
    ]
    
    # Create one branch for each value
    for value in sorted(data[best_feature].unique()):
        
        subset = data[data[best_feature] == value]
        
        tree[best_feature][value] = build_id3_tree(
            subset,
            remaining_features,
            target
        )
    
    return tree
```

```python
target = "ObesityCategory"

features = [
    "Age",
    "Gender",
    "Height",
    "Weight",
    "BMI",
    "PhysicalActivityLevel"
]

id3_tree = build_id3_tree(
    obesity_binary,
    features,
    target
)

id3_tree
```

```python
from graphviz import Digraph
from IPython.display import display, SVG

def visualize_tree(tree):
    
    dot = Digraph()
    
    node_counter = 0
    
    def add_nodes(tree, parent=None, edge_label=None):
        nonlocal node_counter
        
        node_id = f"node{node_counter}"
        node_counter += 1
        
        # Leaf node
        if not isinstance(tree, dict):
            
            dot.node(
                node_id,
                f"Class = {tree}",
                shape="box"
            )
        
        # Internal node
        else:
            feature = next(iter(tree))
            
            dot.node(
                node_id,
                feature
            )
            
            for value, subtree in tree[feature].items():
                
                child_id = add_nodes(
                    subtree,
                    node_id,
                    str(value)
                )
                
                dot.edge(
                    node_id,
                    child_id,
                    label=str(value)
                )
        
        return node_id
    
    add_nodes(tree)
    
    display(SVG(dot.pipe(format="svg")))
```

```python
visualize_tree(id3_tree)
```

Answer ...

An ID3 decision tree was constructed from scratch using entropy and Information Gain. The attribute with the highest Information Gain was selected as the root node. In this dataset, Weight was selected as the root, and the resulting tree was visualized using Graphviz.

Q18 : What is the maximum depth of the tree?

```python
def tree_depth(tree):
    if not isinstance(tree, dict):
        return 0
    
    feature = next(iter(tree))
    
    return 1 + max(
        tree_depth(subtree)
        for subtree in tree[feature].values()
    )

tree_depth(id3_tree)
```

The maximum depth of the tree is 1.

Q19 : How many leaf nodes do you have?

```python
def count_leaves(tree):
    if not isinstance(tree, dict):
        return 1
    
    feature = next(iter(tree))
    
    return sum(
        count_leaves(subtree)
        for subtree in tree[feature].values()
    )

count_leaves(id3_tree)
```
